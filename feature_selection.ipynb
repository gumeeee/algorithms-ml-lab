```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import chi2, SelectKBest
```

```python
anuncio = pd.read_csv('ad.data', header=None)
anuncio.shape
```

```python
X = anuncio.iloc[:,:-1].values
y = anuncio.iloc[:,-1].values
```

```python
X
```

```python
y
```

```python
X_treinamento, X_teste, ytreinamento, y_teste = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
modelo1 = GaussianNB()
modelo1.fit(X_treinamento, ytreinamento)
previsoes1 = modelo1.predict(X_teste)
print(accuracy_score(y_teste, previsoes1))
```

```python
selecao = SelectKBest(chi2, k=7)
X_novo = selecao.fit_transform(X, y)
```

```python
X_novo
```

```python
X_novo.shape
```

```python
print(selecao.get_support())
```

```python
X_treinamento, X_teste, ytreinamento, y_teste = train_test_split(X_novo, y, test_size=0.3, random_state=0)
```

```python
modelo2 = GaussianNB()
modelo2.fit(X_treinamento, ytreinamento)
previsoes2 = modelo2.predict(X_teste)
print(accuracy_score(y_teste, previsoes2))
```
